```python
from datetime import date
class student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    @classmethod
    def calculate_age(cls, name, birthyear):
        return cls(name, date.today().year - birthyear)
    def show(self):
        print(self.name + "'s age is'" + str(self.age))
```

```python
age = student ('VW', 20)
age.show()
```

```python
by = student.calculate_age('vw', 1988)
by.show()
```

```python
# mini project
class BankAccount:
    def __init__(self, name, balance=0):
        # instance attribute
        self.name =name
        self.balance = balance

    # deposit money
    def deposit(self, amount):
        if amount > 0 :
            self.balance += amount
            print("Deposit successfully")
            print("Deposited: ", amount)
        else:
            print("Invalid deposit amount")

    # Withdraw money 
    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")
        elif amount > self.balance:
            print("Insufficient balance")
        else:
            self.balance -= amount
            print("Withdrawal successful")
            print("Withdrawn: ", amount)

    # check balance
    def check_balance(self):
        print("Current balance:", self.balance)

    # display account details
    def display_details(self):
        print("Account holder", self.name)
        print("Balance:", self.balance)
```

```python
acc1 = BankAccount('Neeta', 50000)
```

```python
acc1.display_details()
```

```python
acc1.deposit(500)
```

```python
acc1.display_details()
```

```python
acc1.withdraw(2500)
```

```python
acc1.check_balance()
```

<h3>Inheritance, Encapsulation and Polymorphism</h3> 

```python
# parent class
class Person:
    def person_info(self, name, age):
        print('Inside Person class')
        print('Name:', name, 'Age:', age)
# parent class 2
class Company:
    def company_info(self, company_name, location):
        print('Inside Company class')
        print('Name:', company_name, 'Location:', location)
```

```python
# child class
class Employee(Person, Company):
    def Employee_info(self, salary, skill):
        print('Inside employee class - child class')
        print('Salary:', salary, 'Skills:', skill)
```

```python
# Create an object of Employee
emp2 = Employee()
```

```python
emp2.person_info('VW', 20)
emp2.company_info('Edunet Foundation', 'Bangalore')
emp2.Employee_info(25000,'AIML Engineer')
```

```python
class Vehicle:
    def info(self):
        print('This is Vehicle')
class Car(Vehicle):
    def car_info(self, name):
        print('Car name is :', name)
class Truck(Vehicle):
    def truck_info(self, name):
        print('Truck name is:', name)
```

```python
obj = Car()
obj.info()
obj.car_info('Tesla')
```

```python
obj = Truck()
obj.info()
obj.truck_info('Tesla')
```

```python
# hybrid inheritance
class Vehicle:
    def vehicle_info(self):
        print('Inside Vehicle class')
class Car(Vehicle):
    def car_info(self):
        print('Inside car class')
class Truck(Vehicle):
    def truck_info(self):
        print('Inside truck class')
class SportsCar(Car, Vehicle):
    def sports_car_info(self):
        print('Inside sportscar class')
```

```python
# create an object
s_car = SportsCar()
s_car.vehicle_info()
s_car.car_info()
s_car.sports_car_info()
```

```python
s_car.truck_info()
```

```python
# public member
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    # public instance method
    def show(self):
        # access public data members
        print('Name', self.name, 'Salary', self.salary)
```

```python
emp = Employee('RMS', 10000)
# access the public data members
print('Name', emp.name, 'Salary', emp.salary)
emp.show()
```

```python
# Private member
class Employee:
    def __init__(self, name, salary):
        self.name = name # public data member
        self.salary = salary # private data member
        
```

```python
# create an object
emp = Employee('RMS', 5000)
# access the private data member
print('Salary', emp.__salary)
```

```python
# protected member
class Company:
    def __init__(self):
        #protected member
        self._project = 'NLP'
# Child class
class Employee(Company):
    def __init__(self, name):
        self.name = name
        Company.__init__(self)
    def show(self):
        print("Employee Nmae:", self.name)
        # Access the protected member
        print('Working on project:', self._project)
```

```python
c = Employee('RMS')
c.show()
```

```python
# direct access
print('Project:', c._project)
```

```python
class India():
    def capital(self):
        print("New Delhi is the capital of India")
    def language(self):
        print("India has 22 official languages")
    def type(self):
        print("India is a developing country")
class USA():
    def capital(self):
        print("Capital of USA is Washington DC")
    def language(self):
        print("The official language in USA is English")
    def type(self):
        print("USA is a developed nation")
            
```

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    def move(self):
        print("Drive")
class Boat:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    def move(self):
        print("Sail")
class Plane:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    def move(self):
        print("Fly")
```

```python
ob1.move()
ob2.move()
ob3.move()
```

```python
# Polymorphism with inheritance
class Bird:
    def intro(self):
        print("There are many types of birds")
    def flight(self):
        print('Most of the  birds can fly but some cannot')
class sparrow(Bird):
    def flight(self):
        print("Sparrows can fly")
class Ostrich(Bird):
    def flight(self):
        print("Ostrich cannot fly")
```

```python
ob_bird = Bird()
ob_bird.intro()
```

```python
ob_bird.flight()
```

```python
ob_sparrow = sparrow()
ob_sparrow.flight()
```

```python
ob_ostrich = Ostrich()
ob_ostrich.flight()
```

```python
# overloading method
class complex:
    def __init__(self, a, b):
        self.a = a
        self.b = b
    def __add__(self, other):
        return self.a + other.a, self.b, other.b
ob1 = complex(1,2)
ob2 = complex(2,3)
ob3 = ob1 + ob2
print(ob3)
```
