```python
# __init__ constructor - used to initialize
class person:
    def __init__(self, name, age, school):
        self.name = name
        self.age = age
        self.school = school
    def myfun(self):
        print('Hello, My name is', self.name)
        print('My age is', self.age)
        print('I went to', self.school)
```

```python
p1 = person('Chinki', 8, 'Binki School')
p1.myfun()
```

```python
class Person:
    def __init__(god, name, age):
        god.name = name
        god.age = age
    def myfunc(god):
        print('Namaskar, My name is', god.name)
        print('My age is', god.age)
```

```python
q1 = Person('Krishna', 567)
```

```python
# accessing attributes and methods
class emp:
    name = 'RMS'
    salary = '25000'
    def show(self):
        print(self.name)
        print(self.salary)
```

```python
e1 = emp()
```

```python
# using getattr function
print(getattr(e1, 'name'))
```

```python
# hasattr - return ture if the object has attribute
print(hasattr(e1, 'name'))
```

```python
print(hasattr(e1, 'gender'))
```

```python
# setattr - sets an attribute
setattr(e1, 'gender', 'F')
```

```python
print(getattr(e1, 'gender'))
```

```python
# delattr - deletes the attribute
delattr(e1, 'gender')
```

```python
print(getattr(e1, 'gender'))
```

```python
class student:
    def __init__(self, usn, name, age): # constructor method
        # Instance variable
        self.usn = usn
        self.name = name
        self.age = age
    # instance method instance variable
    def show(self):
        print('USN', self.usn)
        print('Name', self.name)
        print('Age', self.age)
    # instance method to modify instance variable 
    def update(self, usn, age):
        self.usn = usn
        self.age = age
```

```python
# create an object
stu = student(40, 'VW', 20)
stu.show()
```

```python
stu.update(12, 24)
stu.show()
```
